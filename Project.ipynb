# Analytical Derivation of the $e^+ e^- \to \mu^+ \mu^-$ Cross Section

## Matrix Element from Feynman Rules
We consider the unpolarized scattering process $e^+(p_2, s_2) + e^-(p_1, s_1) \to \mu^-(p_3, s_3) + \mu^+(p_4, s_4)$ via a virtual photon mediator in the $s$-channel.

The invariant matrix element $\mathcal{M}$ is given by:
$$\mathcal{M} = \color{purple}{\overline{v}(p_2, s_2)} \color{#D9CD25}{i e \gamma^{\mu}} \color{pink}{u(p_1, s_1)} \left( \color{orange}{-\frac{i \eta_{\mu \nu}}{q^2}} \right) \color{lightblue}{\overline{u}(p_3, s_3)} \color{#D9CD25}{i e \gamma^{\nu}} \color{darkblue}{v(p_4, s_4)}$$

Using the Mandelstam variable $s = q^2 = (p_1 + p_2)^2$, we can simplify the expression:
$$\mathcal{M} = \frac{i e^2}{s} \left[ \overline{v}(p_2, s_2) \gamma^{\mu} u(p_1, s_1) \right] \eta_{\mu \nu} \left[ \overline{u}(p_3, s_3) \gamma^{\nu} v(p_4, s_4) \right]$$

### Complex Conjugate Matrix Element
Using the property $$\left( \bar{v} \gamma^\mu u \right)^* = u^\dagger (\gamma^\mu)^\dagger (\gamma^0)^\dagger v = u^\dagger (\gamma^\mu)^\dagger \gamma^0 v = u^\dagger \gamma^0 \gamma^\mu v = \bar{u} \gamma^\mu v.$$ the complex conjugate matrix element $\mathcal{M}^*$ is:
$$\mathcal{M}^* = -\frac{i e^2}{s} \left[ \overline{u}(p_1, s_1) \gamma^{\rho} v(p_2, s_2) \right] \eta_{\rho \sigma} \left[ \overline{v}(p_4, s_4) \gamma^{\sigma} u(p_3, s_3) \right]$$






## Unpolarized Squared Matrix Element
To find the cross-section, we average over the initial electron/positron spins (factor of $1/4$) and sum over the final muon spins:
$$\begin{aligned}
\langle|\mathcal{M}|^2\rangle &= \frac{1}{4} \sum_{\text{spins}} |\mathcal{M}|^2 \\
&= \frac{1}{4} \sum_{\text{spins}} \frac{e^4}{s^2} \left[ \bar{v}(p_2, s_2) \gamma^\mu u(p_1, s_1) \eta_{\mu\nu} \bar{u}(p_3, s_3) \gamma^\nu v(p_4, s_4) \right] \times \\
&\quad \left[ \bar{u}(p_1, s_1) \gamma^\rho v(p_2, s_2) \eta_{\rho\sigma} \bar{v}(p_4, s_4) \gamma^\sigma u(p_3, s_3) \right]
\end{aligned}$$



Applying:

* The completeness relations:
  * $\sum_{s} u(p,s)\overline{u}(p,s) = \not{p} + m$
  * $\sum_{s} v(p,s)\overline{v}(p,s) = \not{p} - m$

* The relativistic approximation for electrons $m_e \sim 0$
(since the positron beam has very high energy compared to the electron mass ($E \gg m_e$), we treat the electron as massless ($m_e \sim 0$). However, **we keep the muon mass $m_\mu \neq 0$** because the reaction happens close to the threshold ($\sqrt{s} \sim 2m_\mu$))

* The trace theorems:
  1. Odd number of $\gamma$-matrices: $\text{Tr}[\gamma^\mu \gamma^\nu \gamma^\rho] = 0$.
  2. Trace of two $\gamma$-matrices: $\text{Tr}[\gamma^\mu \gamma^\nu] = 4\eta^{\mu\nu}$.
  3. Trace of four $\gamma$-matrices: $\text{Tr}[\gamma^\mu \gamma^\nu \gamma^\rho \gamma^\sigma] = 4(\eta^{\mu\nu}\eta^{\rho\sigma} - \eta^{\mu\rho}\eta^{\nu\sigma} + \eta^{\mu\sigma}\eta^{\nu\rho})$.


The expression simplifies and splits into an electronic tensor and a muonic tensor:
$$\langle|\mathcal{M}|^2\rangle = \frac{1}{4} \frac{e^4}{s^2} \text{Tr}\left[ \not{p}_1 \gamma^{\rho} \not{p}_2 \gamma^{\mu} \right] \text{Tr}\left[ (\not{p}_3 + m_\mu) \gamma^{\nu} (\not{p}_4 - m_\mu) \gamma^{\sigma} \right] \eta_{\mu \nu} \eta_{\rho \sigma}$$

The traces can be computed as follows
1. **Electronic Trace:**
   $$\text{Tr}\left[ \not{p}_1 \gamma^{\rho} \not{p}_2 \gamma^{\mu} \right] = 4 \left( p_1^\rho p_2^\mu - \eta^{\rho \mu}(p_1 \cdot p_2) + p_1^\mu p_2^\rho \right)$$

2. **Muonic Trace:**
   $$\text{Tr}\left[ (\not{p}_3 + m_\mu) \gamma^{\nu} (\not{p}_4 - m_\mu) \gamma^{\sigma} \right] = 4 \left( p_3^\nu p_4^\sigma - \eta^{\nu \sigma}(p_3 \cdot p_4) + p_3^\sigma p_4^\nu - m_\mu^2 \eta^{\nu \sigma} \right)$$

Contracting the Lorentz indices using $\eta_{\mu\nu}\eta_{\rho\sigma}$ yields:
$$\langle|\mathcal{M}|^2\rangle = \frac{8 e^4}{s^2} \left[ (p_1 \cdot p_3)(p_2 \cdot p_4) + (p_1 \cdot p_4)(p_2 \cdot p_3) + m_\mu^2 (p_1 \cdot p_2) \right]$$




## Kinematics in the Center-of-Mass Frame

In the Center-of-Mass (CM) frame, the collision alignment is chosen along the $z$-axis, with the scattering process occurring in the $yz$-plane, as illustrated in the figure below.


Using this coordinate setup, the initial and final four-momenta can be expressed as functions of the center-of-mass energy $\sqrt{s}$ and the scattering angle $\theta$:

* **Initial State ($e^-$, $e^+$):**
  $$p_1 = (E, 0, 0, E), \quad p_2 = (E, 0, 0, -E) \quad \text{where } E = \frac{\sqrt{s}}{2}$$

* **Final State ($\mu^-$, $\mu^+$):**
  $$p_3 = (E, 0, |\vec{p}_f|\sin\theta, |\vec{p}_f|\cos\theta)$$
  $$p_4 = (E, 0, -|\vec{p}_f|\sin\theta, -|\vec{p}_f|\cos\theta)$$

Where the final momentum magnitude is constrained by $E^2 = |\vec{p}_f|^2 + m_\mu^2$:
$$|\vec{p}_f| = \sqrt{E^2 - m_\mu^2} = \sqrt{\frac{s}{4} - m_\mu^2}$$

### Evaluating Four-Vector Dot Products
Using $a \cdot b = a_0 b_0 - \vec{a} \cdot \vec{b}$:
* $(p_1 \cdot p_2) = 2E^2 = \frac{s}{2}$
* $(p_1 \cdot p_3) = (p_2 \cdot p_4) = E^2 - E|\vec{p}_f|\cos\theta$
* $(p_1 \cdot p_4) = (p_2 \cdot p_3) = E^2 + E|\vec{p}_f|\cos\theta$

Substituting these back into our squared matrix element:
$$\langle|\mathcal{M}|^2\rangle = 16 e^4 \left[ \frac{E^4}{s^2}\left( (1 + \frac{m_\mu^2}{E^2}) + (1 - \frac{m_\mu^2}{E^2})\cos^2\theta \right) \right]$$

Expressing everything in terms of $s$:
$$\langle|\mathcal{M}|^2\rangle = e^4 \left[ \left(1 + \frac{4m_\mu^2}{s}\right) + \left(1 - \frac{4m_\mu^2}{s}\right)\cos^2\theta \right]$$



## Differential Cross Section
The two-body differential cross-section in the CoM frame is defined as:
$$\frac{d\sigma}{d\Omega} = \frac{1}{64\pi^2 s} \frac{|\vec{p}_f|}{|\vec{p}_i|} \langle|\mathcal{M}|^2\rangle$$

Note that the velocity parameter $\beta = \frac{|\vec{p}_f|}{E} = \sqrt{1 - \frac{4m_\mu^2}{s}}$

Since $|\vec{p}_i| = E$ and $\frac{|\vec{p}_f|}{|\vec{p}_i|} = \beta$:
$$\frac{d\sigma}{d\Omega} = \frac{e^4}{64\pi^2 s} \sqrt{1 - \frac{4m_\mu^2}{s}} \left[ \left(1 + \frac{4m_\mu^2}{s}\right) + \left(1 - \frac{4m_\mu^2}{s}\right)\cos^2\theta \right]$$

Using the fine-structure constant $\alpha = \frac{e^2}{4\pi}$:
$$\frac{d\sigma}{d\Omega} = \frac{\alpha^2}{4s} \sqrt{1 - \frac{4m_\mu^2}{s}} \left[ \left(1 + \frac{4m_\mu^2}{s}\right) + \left(1 - \frac{4m_\mu^2}{s}\right)\cos^2\theta \right]$$

```python
# --- CONSTANTS ---
m = 105.658 # m_mu (MeV)
m_e = 0.511 # m_el (MeV)
epsilon = 0.05
s1 = ((2. + epsilon) * m)**2
num_events = int(100000) # 10^5 events for Task 5

# --- CROSS SECTION ---
def differential_xs(theta, s):
    alpha = 1/137.036
    hbarc2_to_nb = 3.89379e11
    beta = np.sqrt(1 - 4*m**2/s)
    ds_nat = (alpha**2 / (4 * s)) * beta * ((1 + 4*m**2/s) + (1 - 4*m**2/s) * np.cos(theta)**2)
    return ds_nat * hbarc2_to_nb

def weights(theta, s):
    # Includes the sin(theta) phase space factor
    return differential_xs(theta, s) * np.sin(theta)
```

2. Compute and display the angle and momentum components distributions of the emerging muon pairs;

The momentum components of the outgoing muons are computed in the center-of-mass (CM) frame from the scattering angles $\theta$ and $\phi$.

Since the two muons are produced back-to-back, each carries half of the total center-of-mass energy,

$$
E_\mu = \frac{\sqrt{s}}{2}.
$$

The magnitude of the three-momentum is obtained from the relativistic energy-momentum relation,

$$
E_\mu^2 = |\mathbf{p}|^2 + m_\mu^2,
$$

which gives

$$
|\mathbf{p}| = \sqrt{E_\mu^2 - m_\mu^2}
             = \sqrt{\frac{s}{4} - m_\mu^2}.
$$

The transverse momentum is obtained by projecting the momentum vector onto the plane perpendicular to the beam axis,

$$
p_T = |\mathbf{p}| \sin\theta.
$$

Using the azimuthal angle $\phi$, the transverse momentum is then decomposed into its Cartesian components,

$$
p_x = p_T \cos\phi,
$$

$$
p_y = p_T \sin\phi.
$$

The longitudinal component is

$$
p_z = |\mathbf{p}| \cos\theta.
$$

```python
# --- CM MOMENTUM COMPONENTS ---
def pt_component(s, m, theta):
    return np.sqrt(s/4 - m**2) * np.sin(theta)

def pz_component(s, m, theta):
    return np.sqrt(s/4 - m**2) * np.cos(theta)

def px_component(s, m, theta, phi):
    return pt_component(s, m, theta) * np.cos(phi)

def py_component(s, m, theta, phi):
    return pt_component(s, m, theta) * np.sin(phi)
```

3. Boost muons four-momenta in the laboratory frame, i.e. in the frame where the electron is at rest and the positron has enough energy to give rise to the process;

The momentum components computed in the previous section are expressed in the center-of-mass (CM) frame. To obtain the corresponding quantities in the laboratory frame, a Lorentz boost is applied along the beam ($z$) axis.

Assuming a fixed-target configuration, one electron is initially at rest, with four-momentum

$$
p_1 = (m_e,\mathbf{0}),
$$

while the incoming electron has four-momentum

$$
p_2 = (E_2,\mathbf{p}).
$$

The center-of-mass energy is given by the Mandelstam invariant

$$
s = (p_1+p_2)^2.
$$

Expanding the square,

$$
\begin{aligned}
s &= p_1^2 + p_2^2 + 2p_1\cdot p_2 \\
  &= m_e^2 + m_e^2 + 2m_eE_2,
\end{aligned}
$$

where we used $p_1^2=p_2^2=m_e^2$ and $p_1\cdot p_2=m_eE_2$.

Solving for the beam energy gives

$$
E_2 = \frac{s-2m_e^2}{2m_e},
$$

which is the expression implemented in the code.

The boost parameters are then computed as

$$
\beta=\frac{|p|}{E_{tot}}=\frac{\sqrt{E_2^2-m_e^2}}{E_2+m_e},
$$

and

$$
\gamma=\frac{1}{\sqrt{1-\beta^2}}.
$$

Finally, the Lorentz transformation along the $z$ axis is applied to the energy and longitudinal momentum:

$$
E_{\rm lab}=\gamma(E+\beta p_z),
$$

$$
p_{z,\rm lab}=\gamma(p_z+\beta E).
$$

Since the boost is performed along the beam direction, the transverse momentum components remain unchanged,

$$
p_x^{\rm lab}=p_x^{\rm CM}, \qquad
p_y^{\rm lab}=p_y^{\rm CM}.
$$

```python
# --- LORENTZ BOOST TO LAB FRAME ---
def boost_cm_to_lab(s, m, m_e, E, pz):
    E2 = (s - 2*m_e**2) / (2*m_e)
    beta = np.sqrt(E2**2 - m_e**2) / (E2 + m_e)
    gamma = 1 / np.sqrt(1 - beta**2)

    pz_lab = gamma * (pz + beta * E)
    E_lab = gamma * (E + beta * pz)
    # px, py remain unchanged
    return E_lab, pz_lab
```

4. Write a Monte Carlo simulation that generates scattering events following the distrubtions that you found analytically;

Because the probability of a muon scattering at a specific angle θ is not uniform (it heavily favors forward and backward directions), we cannot simply use a standard random number generator.

To generate discrete physical events from our continuous Quantum Electrodynamics (QED) probability curve, we use the **Rejection sampling Method**.

1. **The Bounding Box (w_max):** We evaluate the theoretical probability curve to find its absolute maximum peak, defining the "height" of our random sampling area.
2. **Throwing Random Darts:** For each loop, we generate a completely random candidate angle ($\theta$) and a random probability check ($r$) between $0$ and $w_{max}$.
3. **The Physics Check:**
   * If $r < \text{weights}(\theta)$, the random point landed *under* the curve. The event is physically valid and **Accepted**.
   * If $r \ge \text{weights}(\theta)$, the point landed in empty space. It is **Rejected** and discarded.

By repeating this process, the areas with higher probabilities naturally accumulate more events.



```python
# --- MONTE CARLO SIMULATION ---
def run_monte_carlo(n_events, s, m):
    theta_test = np.linspace(0, np.pi, 1000)
    w_max = np.max(weights(theta_test, s))

    accepted_theta = []
    accepted_phi = []

    while len(accepted_theta) < n_events:
        theta_cand = np.random.uniform(0, np.pi)
        phi_cand = np.random.uniform(0, 2*np.pi)
        r = np.random.uniform(0, w_max)

        if r < weights(theta_cand, s):
            accepted_theta.append(theta_cand)
            accepted_phi.append(phi_cand)

    return np.array(accepted_theta), np.array(accepted_phi)
```

```python
# --- VISUALIZING THE MONTE CARLO METHOD ---
def plot_accept_reject_demo(s, n_demo_events=2500):
    theta_test = np.linspace(0, np.pi, 1000)
    w_curve = weights(theta_test, s)
    w_max = np.max(w_curve)

    theta_cand = np.random.uniform(0, np.pi, n_demo_events)
    r = np.random.uniform(0, w_max, n_demo_events)

    actual_weights = weights(theta_cand, s)
    accepted_mask = r < actual_weights

    theta_acc = theta_cand[accepted_mask]
    r_acc = r[accepted_mask]

    theta_rej = theta_cand[~accepted_mask]
    r_rej = r[~accepted_mask]

    fig, ax = plt.subplots(figsize=(10, 6), dpi=120)

    ax.axhline(w_max, color='gray', linestyle='--', alpha=0.7, label=f'Bounding Box Height ($w_{{max}}$)')
    ax.plot(theta_test, w_curve, color='black', lw=3, label='Theoretical Distribution $w(\\theta)$')

    ax.scatter(theta_acc, r_acc, color='forestgreen', alpha=0.7, s=15, label=f'Accepted Events ({len(theta_acc)})')
    ax.scatter(theta_rej, r_rej, color='crimson', alpha=0.3, s=15, label=f'Rejected Events ({len(theta_rej)})')

    ax.set_xlim(0, np.pi)
    ax.set_ylim(0, w_max * 1.1)

    ax.set_xticks([0, np.pi/6, np.pi/3, np.pi/2, 2*np.pi/3, 5*np.pi/6, np.pi])
    ax.set_xticklabels([r"$0$", r"$\pi/6$", r"$\pi/3$", r"$\pi/2$", r"$2\pi/3$", r"$5\pi/6$", r"$\pi$"])
    ax.set_xlabel(r'Candidate Scattering Angle $\theta$ (radians)', fontsize=12)
    ax.set_ylabel('Random Probability Check (r)', fontsize=12)
    ax.set_title('Monte Carlo Accept-Reject Method Visualization', fontsize=14, pad=15)
    ax.legend(loc='upper right', framealpha=0.9)

    plt.tight_layout()
    plt.show()

plot_accept_reject_demo(s1, n_demo_events=2500)
```

```python
# ---  GENERATE EVENTS ---
print(f"Generating {num_events} events...")
thetas, phis = run_monte_carlo(num_events, s1, m)

# --- CALCULATE COMPONENTS & BOOST ---
# Muon (-) Components
pz_cm_minus = pz_component(s1, m, thetas)
px_cm_minus = px_component(s1, m, thetas, phis)
py_cm_minus = py_component(s1, m, thetas, phis)
pt_cm_minus = pt_component(s1, m, thetas)
E_cm = np.full_like(thetas, np.sqrt(s1)/2)

# Muon (+) Components (Equal and opposite in CM frame)
pz_cm_plus = -pz_cm_minus
px_cm_plus = -px_cm_minus
py_cm_plus = -py_cm_minus

# Boost BOTH muons to Lab Frame
E_lab_minus, pz_lab_minus = boost_cm_to_lab(s1, m, m_e, E_cm, pz_cm_minus)
E_lab_plus, pz_lab_plus = boost_cm_to_lab(s1, m, m_e, E_cm, pz_cm_plus)
theta_lab_minus = np.arctan2(pt_cm_minus, pz_lab_minus)
theta_lab_plus = np.arctan2(pt_cm_minus, pz_lab_plus)

# --- SYNTHETIC DATASET ---
# We only need the 3 spatial coordinates per muon in the lab frame
df_events = pd.DataFrame({
    "mu_minus_px": px_cm_minus,
    "mu_minus_py": py_cm_minus,
    "mu_minus_pz": pz_lab_minus,
    "mu_plus_px": px_cm_plus,
    "mu_plus_py": py_cm_plus,
    "mu_plus_pz": pz_lab_plus
})




####### CHECK #######

# Invariant mass squared of mu+
p2_lab_plus = px_cm_plus**2 + py_cm_plus**2 + pz_lab_plus**2
m2_lab_plus = E_lab_plus**2 - p2_lab_plus

p2_lab_minus = px_cm_minus**2 + py_cm_minus**2 + pz_lab_minus**2
m2_lab_minus = E_lab_minus**2 - p2_lab_minus

print(f"Expected m² = {m**2:.6f}")
print(f"Mean reconstructed m² (mu-) = {np.mean(m2_lab_minus):.6f}")
print(f"Mean reconstructed m² (mu+) = {np.mean(m2_lab_plus):.6f}")

print(f"Max deviation (mu-) = {np.max(np.abs(m2_lab_minus - m**2)):.3e}")
print(f"Max deviation (mu+) = {np.max(np.abs(m2_lab_plus - m**2)):.3e}")


# boost
diff_minus = E_lab_minus - df_events['mu_minus_pz']
diff_plus = E_lab_plus - df_events['mu_plus_pz']

print('mu plus')
print(f"Max deviation (E - pz): {np.max(diff_plus):.2f} MeV")
print(f"Minimum deviation (E - pz): {np.min(diff_plus):.2f} MeV")
print(f"Mean deviation  (E - pz): {np.mean(diff_plus):.2f} MeV")

print('mu minus')
print(f"Max deviation (E - pz): {np.max(diff_minus):.2f} MeV")
print(f"Minimum deviation (E - pz): {np.min(diff_minus):.2f} MeV")
print(f"Mean deviation  (E - pz): {np.mean(diff_minus):.2f} MeV")
```

## Angular Distributions

The simulated angular distributions are compared with the theoretical expectations to validate the Monte Carlo event generation.

### Azimuthal Angle Distribution

The azimuthal angle $\phi$ is generated from a uniform distribution in the interval $[0,2\pi]$, reflecting the cylindrical symmetry of the scattering process around the beam axis. Consequently, its probability density is

$$
f(\phi)=\frac{1}{2\pi},
\qquad 0\leq\phi<2\pi.
$$

The normalized histogram of the simulated events is therefore expected to be flat and is compared with the corresponding constant theoretical prediction.

### Scattering Angle Distribution

The polar scattering angle $\theta$ is not uniformly distributed. Instead, it follows the probability density derived from the QED differential cross section,

$$\frac{d\sigma}{d\Omega} = \frac{\alpha^2}{4s} \sqrt{1 - \frac{4m_\mu^2}{s}} \left[ \left(1 + \frac{4m_\mu^2}{s}\right) + \left(1 - \frac{4m_\mu^2}{s}\right)\cos^2\theta \right]$$

Since the events are generated in terms of the polar angle $\theta$, the probability density also includes the Jacobian associated with the solid-angle element,

$$
d\Omega=\sin\theta\,d\theta\,d\phi,
$$

so that

$$
P(\theta) =
\frac{d\sigma}{d\Omega}\sin\theta.
$$

The theoretical distribution is evaluated using the `weights` function and normalized numerically so that its integral over the interval is equal to one. This normalization allows a direct comparison with the histogram, which is also normalized by setting `density=True`.

```python
# --- PLOTTING ANGULAR DISTRIBUTIONS ---
fig, axs = plt.subplots(1, 2, figsize=(14, 5), dpi=100)

# Phi Distribution (The Azimuthal Angle)
# It should be perfectly flat since we used a uniform random generator
axs[0].hist(phis, bins=50, histtype='stepfilled', color='darkorange', edgecolor='black', alpha=0.5, density=True, label='Simulated Data')
# The theoretical line for a uniform distribution from 0 to 2*pi is just a horizontal line at 1/(2*pi)
axs[0].axhline(1 / (2 * np.pi), color='red', linestyle='--', linewidth=2.5, label='Theoretical (Uniform)')

axs[0].set_title(r'Azimuthal Angle ($\phi$) Distribution')
axs[0].set_xlabel(r'$\phi$ (radians)')
axs[0].set_ylabel('Probability Density')
axs[0].set_xlim(0, 2*np.pi)
axs[0].legend(loc='upper right')
axs[0].set_xticks([0, np.pi/3, 2*np.pi/3, np.pi, 4*np.pi/3, 5*np.pi/3, 2*np.pi])
axs[0].set_xticklabels([r"$0$", r"$\pi/3$", r"$2\pi/3$", r"$\pi$", r"$4\pi/3$", r"$5\pi/3$", r"$2\pi$"])
axs[0].minorticks_on()
axs[0].tick_params(axis='both', which='both', direction='in',
                top=True, right=True, labelsize=12)
axs[0].tick_params(axis='both', which='minor', length=4)
axs[0].tick_params(axis='both', which='major', length=7)

# Theta Distribution (The Scattering Angle)
# Plot the histogram of our simulated thetas (normalized to an area of 1)
axs[1].hist(thetas, bins=50, histtype='stepfilled', color='steelblue', edgecolor='black', alpha=0.5, density=True, label='Monte Carlo Data')

# Calculate the Theoretical Curve
theta_line = np.linspace(0, np.pi, 500)
# We use your 'weights' function which already includes the sin(theta) factor!
y_th = weights(theta_line, s1)
# To compare a mathematical function to a 'density=True' histogram,
# we must normalize the theoretical curve so the area under it equals 1.
# np.trapz numerically integrates the area under the curve so we can divide by it.
y_th_normalized = y_th / np.trapezoid(y_th, theta_line)

# Plot the theoretical curve
axs[1].plot(theta_line, y_th_normalized, color='red', linewidth=2.5, linestyle='-', label='Theoretical QED PDF')

axs[1].set_title(r'Scattering Angle ($\theta$) Distribution')
axs[1].set_xlabel(r'$\theta$ (radians)')
axs[1].set_ylabel('Probability Density')
axs[1].set_xlim(0, np.pi)
axs[1].legend(loc='upper right')
axs[1].set_xticks([0, np.pi/6, np.pi/3, np.pi/2, 2*np.pi/3, 5*np.pi/6, np.pi])
axs[1].set_xticklabels([r"$0$", r"$\pi/6$", r"$\pi/3$", r"$\pi/2$", r"$2\pi/3$", r"$5\pi/6$", r"$\pi$"])
axs[1].minorticks_on()
axs[1].tick_params(axis='both', which='both', direction='in',
                top=True, right=True, labelsize=12)
axs[1].tick_params(axis='both', which='minor', length=4)
axs[1].tick_params(axis='both', which='major', length=7)

plt.tight_layout()
plt.show()
```

## Kinematic Distributions in the CM and Laboratory Frames

The following plots illustrate the distributions of the momentum components and scattering angle obtained from the Monte Carlo simulation. These quantities are determined by the scattering angle $\theta$ and the center-of-mass energy $\sqrt{s}$.

### Transverse Momentum in the CM Frame

The transverse momentum is given by

$$
p_T = |\mathbf{p}|\sin\theta,
$$

where

$$
|\mathbf{p}|=\sqrt{\frac{s}{4}-m_\mu^2}.
$$

Since $\sin\theta$ reaches its maximum at $\theta=\pi/2$, the transverse momentum is largest for particles emitted perpendicular to the beam axis and vanishes for forward ($\theta=0$) and backward ($\theta=\pi$) scattering.

Increasing the center-of-mass energy increases the momentum magnitude $|\mathbf{p}|$, causing the entire $p_T$ distribution to extend towards larger values.

### Longitudinal Momentum in the CM Frame

The longitudinal momentum is

$$
p_z = |\mathbf{p}|\cos\theta.
$$

Because the scattering process is symmetric under $\theta \rightarrow \pi-\theta$, the distribution is symmetric around $p_z=0$, with positive and negative values corresponding to the two opposite emission directions of the muons.

As $\sqrt{s}$ increases, the maximum value of $|p_z|$ also increases proportionally to the momentum magnitude.

### Scattering Angle in the Laboratory Frame

After applying the Lorentz boost along the beam axis, the scattering angle is reconstructed from the boosted momentum components. The boost leaves the transverse momentum unchanged while increasing the longitudinal momentum according to

$$
p_{z,\mathrm{lab}}=\gamma(p_z+\beta E).
$$

As a consequence, the emitted particles become more collimated along the boost direction, and the laboratory angular distribution is shifted towards smaller values of $\theta$.

### Longitudinal Momentum in the Laboratory Frame

Unlike the CM distribution, the laboratory longitudinal momentum is no longer symmetric around zero. The Lorentz boost shifts the distribution towards positive values, reflecting the motion of the center-of-mass frame with respect to the laboratory.

A larger boost (larger $\beta$) produces a stronger asymmetry and higher laboratory momenta.

## Event Weight Calculation

The Monte Carlo sample contains an arbitrary number of simulated events and therefore does not correspond to the number of interactions expected in a real experiment. To relate the simulation to realistic experimental conditions, a rescaling factor is computed.



### Expected Number of Events

The expected number of interactions is computed from

$$
N_{\rm exp}
=
N_{e^+}\,n_e\,\sigma_{\rm tot},
$$

where

- $N_{e^+}= R \times t_{week} $ is the total number of positrons delivered in one week, with a rate $R=10^6 \rm {Hz}$,
- $n_e$ is the target electron areal density,
- $\sigma_{\rm tot}$ is the total cross section, evaluated as 
$$
\sigma_{\rm tot}
=
\int_0^\pi
\frac{d\sigma}{d\Omega}
\,2\pi\sin\theta\,d\theta.
$$

The electron areal density of the 3 cm thick beryllium target is calculated from

$$
n_e
=
\frac{\rho N_A}{A}\,Z\,L,
$$

where $\rho$ is the material density, $A$ its molar mass, $Z$ the number of electrons per atom, $L$ the target thickness, and $N_A$ Avogadro's constant.

### Event Weight

Since the simulation contains $N_{\rm sim}$ generated events, each event is assigned a weight

$$
w
=
\frac{N_{\rm exp}}{N_{\rm sim}},
$$

so that the weighted Monte Carlo sample represents the statistics expected after one week of continuous data taking with a positron rate of $10^6$ Hz.

## Task 8: Nominal Beam Energy

The nominal beam energy is determined by requiring that the positron beam retains enough energy at the end of the target to produce a $\mu^+\mu^-$ pair.

### Kinematic Threshold Energy

The production threshold is reached when the center-of-mass energy is equal to the total rest energy of the final-state particles:

$$
\sqrt{s_{\rm th}} = 2m_\mu,
$$

therefore,

$$
s_{\rm th}=(2m_\mu)^2.
$$

For a fixed-target configuration, the invariant mass is approximately

$$
s \simeq 2m_eE_{\rm beam},
$$

where the electron mass of the target is retained while the positron mass is neglected. Solving for the minimum beam energy required at the target exit gives

$$
E_{\rm th}
=
\frac{s_{\rm th}}{2m_e}.
$$

This is the minimum positron energy needed after crossing the target to still allow $\mu^+\mu^-$ production.

### Energy Loss in the Target

While travelling through the beryllium target, the positron beam loses energy due to interactions with the material. The average energy loss is described using the radiation length $X_0$:

$$
E(z)=E_0 e^{-z/X_0},
$$

where

- $E_0$ is the initial beam energy,
- $E(z)$ is the energy after travelling a distance $z$ inside the target,
- $X_0$ is the radiation length of the material.

To ensure that the beam reaches the production threshold at the end of the target ($z=z_{\rm max}$), the required initial energy is obtained by inverting the relation:

$$
E_0 = E_{\rm th} e^{z_{\rm max}/X_0}.
$$
Using the radiation length of beryllium and a target thickness of 3 cm, this gives the nominal starting beam energy required for the experiment.

```python
# --- TASK 8: NOMINAL BEAM ENERGY ---
X0_Be = 35.28 # Radiation length of Beryllium in cm
z_max = 3.0 # Thickness of target in cm

# 1. Kinematic Threshold Energy
s_threshold = (2 * m)**2 # m = 105.658 MeV
# Using s = 2 * m_e * E_beam + m_e^2 + m_pos^2 ≈ 2 * m_e * E_beam
E_threshold = s_threshold / (2 * m_e) # in MeV

# 2. Back-calculate initial energy E0
# E(z) = E0 * exp(-z/X0)  =>  E0 = E(z) * exp(z/X0)
E_nominal = E_threshold * np.exp(z_max / X0_Be)

print(f"Minimum required energy at target exit: {E_threshold / 1000:.2f} GeV")
print(f"Nominal starting beam energy (E0): {E_nominal / 1000:.2f} GeV")
```

### Task 7: Realistic Beam Shape and Energy Spread

Up until now, we have assumed an ideal, perfectly thin beam where every positron has the exact same energy. To adapt our simulation to the real world, we must model the **Transverse Profile** and the **Energy Spread** of the beam.

#### Transverse Beam Profile (Uniform Circle)
We assume the beam hits the target as a flat, uniform circle with a radius of **1 cm**.
To correctly generate random points with a *uniform density* across a circle, we cannot simply pick a random radius $r$. Because the area of a circle grows as $r^2$, a purely uniform $r$ would unphysically clump particles in the center.
Instead, we sample the *area* uniformly by taking the square root of a random number:
$$r = R_{beam} \sqrt{\text{random}(0,1)}$$

#### Beam Energy Spread (Gaussian)
Real accelerators have slight energy fluctuations. We model the positron's starting energy as a **Normal (Gaussian) Distribution**, centered at our calculated nominal energy ($E_{nominal}$) with a standard deviation of **0.5 GeV**.

These physical starting conditions ($x$, $y$, and starting energy) are appended directly to our final synthetic dataset.

```python
# --- TASK 7: BEAM SHAPE AND ENERGY SPREAD ---
# 1. Transverse Profile (Flat circle of R = 1 cm)
# To get a uniform circle, we sample radius with a square root, and angle uniformly
R_beam = 1.0
r_random = R_beam * np.sqrt(np.random.uniform(0, 1, num_events))
theta_beam = np.random.uniform(0, 2*np.pi, num_events)

# XY initial generation coordinates
x_start = r_random * np.cos(theta_beam)
y_start = r_random * np.sin(theta_beam)

# 2. Energy Spread (Gaussian width of 0.5 GeV)
energy_width_MeV = 0.5 * 1000 # Convert to MeV
E_beam_initial = np.random.normal(E_nominal, energy_width_MeV, num_events)

# Add these initial conditions to our synthetic dataset!
df_events['pos_start_x'] = x_start
df_events['pos_start_y'] = y_start
df_events['E_beam_start'] = E_beam_initial

# Save to file
filename = "muon_events_dataset.csv"
df_events.to_csv(filename, index=False)
print(f"Dataset successfully saved to {filename}!")

print("Beam profile and energy spread added to the DataFrame.")
```

## Task 9 (Optional): Z-Dependent Monte Carlo Simulation

In the previous simulations, the beam energy was assumed to be constant throughout the target. In a realistic fixed-target experiment, the incoming positron loses energy while travelling through the beryllium target as seen before ($E(z)=E_0 e^{-z/X_0}$). Therefore, the available center-of-mass energy depends on the interaction point $z$.

The local center-of-mass energy is therefore recalculated for every generated interaction point:

$$
s(z)\simeq 2m_eE(z).
$$

### Generation of Events

For each Monte Carlo event, random values of the scattering angles and production position are generated:

$$
\theta\in[0,\pi],
$$

$$
\phi\in[0,2\pi],
$$

$$
z\in[0,z_{\rm max}].
$$

The event is then accepted according to the local differential cross section evaluated at the corresponding value of $s(z)$.

### Kinematic Threshold

Not every point inside the target can produce a $\mu^+\mu^-$ pair. The reaction is allowed only if the available center-of-mass energy is above threshold:

$$
s(z)\geq 4m_\mu^2.
$$

Events generated in regions where this condition is not satisfied are rejected.

### Importance of the Local Value of $s$

Since the interaction can occur at different depths inside the target, each event is generated with a different value of $s$. Consequently, the Lorentz boost from the CM frame to the laboratory frame cannot use a single fixed value of $s$.

For each event, the boost parameters must therefore be calculated using the corresponding local value:

$$
s \rightarrow s(z).
$$

This accounts for the variation of the beam energy along the target and provides a more realistic description of the experimental conditions.

```python
# --- Z-DEPENDENT MONTE CARLO ---

def run_advanced_monte_carlo(n_events, E0, z_max, X0, m, m_e):
    # E0: Initial beam energy (from Task 8)

    accepted_theta = []
    accepted_phi = []
    accepted_z = []
    accepted_s = [] # We need to save the local 's' for the Lorentz boost later!

    # Find the absolute maximum weight for the Accept-Reject bounding box
    # The cross-section is highest at the front of the target (z=0) where energy is max
    s_max = 2 * m_e * E0
    theta_test = np.linspace(0, np.pi, 1000)
    w_max = np.max(weights(theta_test, s_max))

    print("Running Z-Dependent Monte Carlo...")

    while len(accepted_theta) < n_events:
        # Generate random candidates
        theta_cand = np.random.uniform(0, np.pi)
        phi_cand = np.random.uniform(0, 2*np.pi)
        z_cand = np.random.uniform(0, z_max)
        r = np.random.uniform(0, w_max)

        # Calculate local energy and s at this depth z
        E_local = E0 * np.exp(-z_cand / X0)
        s_local = 2 * m_e * E_local

        # Kinematic Threshold Check
        if s_local < 4 * m**2:
            continue # Energy too low to create muons, reject and try again

        # Cross-Section Accept/Reject (using local s)
        if r < weights(theta_cand, s_local):
            accepted_theta.append(theta_cand)
            accepted_phi.append(phi_cand)
            accepted_z.append(z_cand)
            accepted_s.append(s_local)

    return np.array(accepted_theta), np.array(accepted_phi), np.array(accepted_z), np.array(accepted_s)

# --- EXECUTE TASK 9 ---
# (Using the E_nominal calculated in Task 8)
# We generate a smaller dataset here just to prove it works, as it takes slightly longer
n_advanced_events = 100000
thetas_adv, phis_adv, z_adv, s_adv = run_advanced_monte_carlo(n_advanced_events, E_nominal, z_max, X0_Be, m, m_e)

# NOTE: When doing the Lorentz Boost for these events, you MUST use the
# 's_adv' array instead of a constant 's1', because every particle now has
# a slightly different boost factor depending on where it was created!
```

### Z-Dependent Event Weighting

In this step, we updated our simulation to account for the z dependent weight.
To correctly normalize the 100,000 simulated events to the physical expectations (for 1 week of data taking), we cannot use a constant cross-section as it was done beafore. Instead, we integrated the local cross-section $\sigma(z)$ over the entire target thickness. The calculated weight (approx. **4.69**) dictates how many real-world muon pairs each generated Monte Carlo event represents.
The histogram below visually confirms our physical model:

  - Decay: The production rate peaks at the front face ($z=0$) where the beam energy is at its maximum nominal value. The probability drops exponentially as the beam penetrates deeper and loses energy.

  - Monte Carlo vs. Expected Yield:  The blue histogram shows the raw statistics generated by the Accept-Reject algorithm. The orange histogram applies our scaling weight, mapping the simulation to the actual expected physical yield of nearly 470,000 muon pairs.
